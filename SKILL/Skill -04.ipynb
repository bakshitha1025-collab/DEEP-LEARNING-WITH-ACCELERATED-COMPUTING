```python
%%writefile git_info.py

import subprocess

def show_git_info():

    print("\n--- GIT REPOSITORY INFORMATION ---")

    try:
        branch = subprocess.check_output(
            ["git", "branch", "--show-current"],
            text=True
        ).strip()

        print("Current Branch:", branch)

    except Exception:
        print("Git repository not initialized.")
```

```python
%%writefile pytorch_test.py

import torch

def run_pytorch_test():

    print("\n--- PYTORCH AND CUDA TEST ---")

    if torch.cuda.is_available():

        device = torch.device("cuda")

        print("CUDA Status: Available")
        print("GPU:", torch.cuda.get_device_name(0))

    else:

        device = torch.device("cpu")

        print("CUDA Status: Not Available")
        print("Using CPU")

    a = torch.tensor([10, 20, 30, 40], device=device)

    b = torch.tensor([1, 2, 3, 4], device=device)

    result = a + b

    print("Tensor A:", a)
    print("Tensor B:", b)
    print("Addition Result:", result)

    print("Execution Device:", device)
```
